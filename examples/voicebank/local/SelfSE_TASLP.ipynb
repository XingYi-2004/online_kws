```python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
%config InlineBackend.figure_format = 'retina'

fig_dir = '../exp/paper/SelfSE/fig/'
fontsize = 14

# colors = ['#ff7f0e', '#2ca02c'] # '#ff7f0e'
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b', '#e377c2', '#7f7f7f', '#bcbd22', '#17becf']
```

```python
# 数据
# PESQ = [2.322, 2.296, 2.377, 2.314, 2.292, 2.362, 2.504, 2.554, 2.535, 2.534, 2.397]
# SI_SNR = [14.206, 13.872, 14.619, 14.925, 15.258, 15.088, 17.703, 18.218, 17.559, 18.112, 18.007]

# PESQ = [2.345, 2.27, 2.336, 2.36, 2.385, 2.388, 2.339, 2.381, 2.45, 2.411, 2.395]
# SI_SNR = [14.229, 13.954, 14.146, 14.28, 14.769, 14.847, 15.308, 16.378, 17.285, 17.634, 17.75]

# PESQ = [2.271,2.279,2.329,2.363,2.395,2.377,2.42,2.393,2.484,2.462,2.479]
# SI_SNR = [13.012,13.353,13.884,13.576,14.356,13.992,15.059,15.348,18.674,17.096,18.116]

PESQ = [2.292,2.31,2.351,2.385,2.383,2.348,2.452,2.421,2.481,2.483,2.467]
SI_SNR = [13.003,13.822,14.169,13.939,14.486,14.104,14.987,16.309,18.06,18.035,17.865]

# 横坐标范围
x = np.linspace(0, 1.0, len(PESQ))

# 设置图形大小和字体大小
plt.figure(figsize=(8, 4))
plt.rcParams.update({'font.size': fontsize}) # , 'font.family': 'Times New Roman'

# 创建主轴（左边坐标轴）
ax1 = plt.gca()  # 获取当前轴
line1, = ax1.plot(x, PESQ, marker='o', linestyle='-', color=colors[0], label='PESQ')
ax1.set_xlabel('Sampling Parameter $p$', fontsize=fontsize)
ax1.set_ylabel('PESQ Score', fontsize=fontsize, color=colors[0])
ax1.tick_params(axis='y', labelcolor=colors[0])  # 设置左边坐标轴的颜色

# 创建次轴（右边坐标轴）
ax2 = ax1.twinx()  # 创建共享 X 轴的次轴
line2, = ax2.plot(x, SI_SNR, marker='s', linestyle='--', color=colors[1], label='SI-SNR')
ax2.set_ylabel('SI-SNR (dB)', fontsize=fontsize, color=colors[1])
ax2.tick_params(axis='y', labelcolor=colors[1])  # 设置右边坐标轴的颜色

# 添加图例
# ax1.legend(loc='upper left', fontsize=fontsize)
# ax2.legend(loc='upper left', fontsize=fontsize)
legend_lines = [line1, line2]
legend_labels = [line.get_label() for line in legend_lines]
ax1.legend(legend_lines, legend_labels, loc='upper left', ncol=2, fontsize=fontsize)
ax1.grid(True, linestyle='-.', alpha=0.5) # -- -. :

plt.tight_layout()
plt.savefig(fig_dir + 'study_p.png', dpi=500, bbox_inches='tight')
plt.show()
```

```python
# 数据
PESQ_super = [2.625] * 5
# PESQ = [2.129,2.095,2.168,2.348,2.469,2.129,2.359,2.274,2.293,2.452,2.398,2.459,2.352,2.596,2.504]
PESQ = [2.144,2.151,2.168,2.348,2.469,2.129,2.359,2.274,2.293,2.452,2.398,2.444,2.419,2.575,2.578]

# 横坐标范围
x_labels = ['-5~0dB', '0~5dB', '5~10dB', '10~15dB', '15~20dB']
x = np.arange(len(x_labels))

# 设置图形大小和字体大小
plt.figure(figsize=(8, 4))
plt.rcParams.update({'font.size': 14})

# 绘制折线图
plt.plot(x, PESQ_super, marker='*', linestyle=':', color=colors[4], label='Supervised')
plt.plot(x, PESQ[:5], marker='o', linestyle='-.', color=colors[0], label='NyTT')
plt.plot(x, PESQ[5:10], marker='s', linestyle='--', color=colors[2], label='IDR-SE')
plt.plot(x, PESQ[10:], marker='^', linestyle='-', color=colors[1], label='SelfSE')

# 添加标题和标签
plt.xlabel('SNR Range (dB)')
plt.ylabel('PESQ Score')

# 设置横坐标刻度标签
plt.xticks(x, x_labels, rotation=0)  # 旋转标签以便更清晰显示
# 添加图例
plt.legend()

# 显示网格
plt.grid(True, linestyle='-.', alpha=0.5)  # 使用虚线网格，透明度为 0.5

plt.tight_layout()
plt.savefig(fig_dir + 'SNR_range_result1.png', dpi=500, bbox_inches='tight')
plt.show()
```
