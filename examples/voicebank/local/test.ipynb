```python
import torch
import matplotlib.pyplot as plt
%config InlineBackend.figure_format = 'retina'

# 生成x轴的数据，从1到100（包含1和100），步长为0.1
x = torch.arange(1e-4, 1e-2, 1e-4)

# 计算y轴的数据
y_log2 = torch.log2(x)
y_log = torch.log(x)

# 创建图形和轴
plt.figure(figsize=(10, 6))
plt.plot(x.numpy(), y_log2.numpy(), label='log2(x)')
plt.plot(x.numpy(), y_log.numpy(), label='log(x)')

# 设置标题和标签
plt.title('Comparison of log2 and log functions')
plt.xlabel('x')
plt.ylabel('Value')
plt.legend()
plt.grid(True)

# 显示图形
plt.show()
```
